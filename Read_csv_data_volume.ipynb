```sql
%python

from pyspark.sql.functions import current_timestamp

df_product = ((spark.read.format("csv"))
      .option('header','True')
      .load("/Volumes/fmcg_project/bronze/source_volume_project/productscsv.csv"))

#df_product.show()

df_cust = ((spark.read.format("csv"))
      .option('header','True')
      .load("/Volumes/fmcg_project/bronze/source_volume_project/customerscsv.csv"))

#df_cust.show()

df_aproduct=df_product.withColumn("ingest_datetime",current_timestamp())

df_aproduct.show()

df_acust=df_cust.withColumn("ingest_datetime",current_timestamp())

df_acust.show()

(df_acust.write
.format("delta")
.mode("append")
.saveAsTable("fmcg_project.bronze.acust"))

(df_aproduct.write
.format("delta")
.mode("append")
.saveAsTable("fmcg_project.bronze.aproduct"))
```
