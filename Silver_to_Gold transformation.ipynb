```sql

%python
from pyspark.sql.functions import current_timestamp
from pyspark.sql.functions import count

df_gold = spark.read.table("fmcg_project.silver.acust_silver")

df=df_gold.withColumn("ingest_datetime_gold",current_timestamp())

df_grouped = df.groupBy("country").agg(
    count("*").alias("Total_Count"))

(df_grouped.write
.format("delta")
.mode("overwrite")
.option("overwriteSchema", "true")
.saveAsTable("fmcg_project.gold.Customer_aggregated_data"))

df_grouped.show()
```
