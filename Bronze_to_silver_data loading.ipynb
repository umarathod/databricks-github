```sql
%python
from pyspark.sql.functions import current_timestamp

df_silver = spark.read.table("fmcg_project.bronze.acust")

df=df_silver.withColumn("ingest_datetime_silver",current_timestamp())

(df.write
.format("delta")
.mode("append")
.saveAsTable("fmcg_project.silver.acust_silver"))

df.show()
```
